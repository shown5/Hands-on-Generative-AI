10.3 Mixture of Experts(MoE)

大規模言語モデルの分野における有力なアプローチとして近年注目を集めているのが MoE。
Mistral チームからリリースされており、Transformer と比較して訓練の効率や実運用環境でのスケーラビリティに優れているとされている。
MoE を使うことで効率よく学習曲線を伸ばすことができ、推論時も多数のリクエストを効率的に処理できる。（セルフアテンションの後、順伝搬の前にゲーティング機構があるから？）

MoE では Transformer ブロックの一部または全てを MoE ブロックに置き換える点が特徴。このMoEブロックが複数のエキスパートネットワークから構成されており、各エキスパートは別個のモデルになっている。
トークンが与えられると、ゲーティング機構により適したブロックに振り分けられる構成になっている。この振り分けもゲーティング機構自体が学習する。

大容量な GPU が必要になるためローカルでの推論には向かないが、大規模な運用環境では非常に有効。

10.4 最適化と量子化

高速化技術に Speculative Decoding （推論的でコーディング、アシスト生成）がある。これは同じタスクに対して同じアーキテクチャーのモデルを２つ用意する。一つは小型で高速、もう一つは大型で高品質なモデル。小型モデルで複数トークンを生成し、これを大型モデルに渡す手法。

似た手法に Medusa Decoding がある。これは既存のモデルに新たなヘッドを追加して、一度に複数のトークンを生成できるようにするファインチューニング手法。
