7.3 LoRA の訓練

フルファインチューニングや DreamBooth の悩みは、元の StableDiffusion モデルと同程度のサイズの新しい重みができてしまうこと。
このような場合は LoRA を活用できる。

おさらいになるが、LoRA は事前訓練済みモデルの重みを凍結し、ランク分解行列を挿入することで訓練が必要なパラメーター数を大幅に削減する。

```python
from diffusers import DiffusionPipeline
from huggingface_hub import model_info

#クラシックな手描き漫画風スタイルを使用する
lora_model_id = "alvdansen/littletinies"

info = model_info(lora_model_id)
base_model_id = info.card_data.base_model

pipe = DiffusionPipeline.from_pretrained(
    base_model_id,
    torch_dtype=torch.float16,
)
pipe = pipe.to(device)

pipe.load_lora_weights(lora_model_id)

pipe.fuse_lora()
```

```python
image = pipe(
    "A llama drinking boba tea", num_inference_steps=25, guidance_scale=7.5
).images[0]

image
```
